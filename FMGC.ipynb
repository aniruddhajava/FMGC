```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\ABC\OneDrive\Desktop\py\FMCG Assignment - 2\Raw Sales Data.csv")
df
```

```python
df.to_csv("Raw Sales Data.csv", index=False)
```

```python
df["Month"] = pd.to_datetime(df["Month"], format="%d-%b-%y")
```

```python
print(df["Sales"].dtype)
```

```python
df["Sales"] = df["Sales"].astype(str).str.replace(",", "").astype(int)
```

```python
df["Qty"] = df["Qty"].astype(str).str.replace(",", "").astype(int)
```

```python
df.info()
```

#### 1. Monthwise number of quantities sold


```python
monthly_qty = df.groupby(by=["Month"])[["Qty"]].sum()
monthly_qty
```

#### 2. Category wise Qty sold

```python
category_qty = df.groupby(by=["Category"])[["Qty"]].sum()
category_qty
```

#### 3. Sub-channel wise total sales

```python
sub_Tsales = df.groupby(by=["Sub-Channels"])[["Sales"]].sum()
sub_Tsales
```

#### 4. Highest sale in which month

```python
high_sales_month = df.groupby(by=["Month"])[["Sales"]].sum().sort_values(by="Sales", ascending=False).head(1)
high_sales_month
```

#### 5. Highest selling product (based on sales) (packaging quantity doesn't matter. Ex. consider "GET SLIM
#### JUICE 1 L" and "GET SLIM JUICE 500 ML" as one product)


```python
df["Product"] = df["Product Name"].str.replace(r"\s\d+\s?(ML|L|KG|G)", "", regex=True)
top_product = pd.DataFrame(df.groupby("Product")["Sales"].sum().sort_values(ascending=False).head(1))
top_product
```

#### 6. List of different products company is offering (packaging quantity doesn't matter. Ex. consider "GET
#### SLIM JUICE 1 L" and "GET SLIM JUICE 500 ML" as one product)

```python
df["Product_Name"] = df["Product"].str.replace(r"\s\d+\s?(ML|L|KG|G)","",regex=True)
unique_product = pd.DataFrame(df["Product_Name"].unique(),columns=["Different Products"])
unique_product
```

```python
with pd.ExcelWriter("Sales_Report.xlsx") as writer:

    # Sheet 1
    monthly_qty.to_excel(writer, sheet_name="Monthly Qty")
    #Sheet 2
    category_qty.to_excel(writer, sheet_name="Category Qty")
    #Sheet 3
    sub_Tsales.to_excel(writer, sheet_name="Top Sales")
    #Sheet 4
    high_sales_month.to_excel(writer, sheet_name="High Sales Month")
    #Sheet 5
    top_product.to_excel(writer, sheet_name="Top Product")
    #Sheet 6
    unique_product.to_excel(writer, sheet_name="All Product")
```
